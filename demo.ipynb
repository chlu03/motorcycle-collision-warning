# Motorcycle Collision Warning System — Demo

This notebook runs the full pipeline on one video and shows the results.
The same pipeline can be run from the command line with `python main.py --source videos/CAR.mp4`.

Steps: lane detection → YOLOv5 vehicle detection → lane filtering → DeepSORT tracking →
speed estimation → time-to-collision (TTC) risk assessment → overlay and warning sound.

```python
from src.pipeline import run

outputs = run('videos/CAR.mp4')
```

## Warnings

Times when a vehicle in the rider's lane had a TTC at or below the threshold.

```python
import pandas as pd

warnings = pd.read_csv(outputs['warnings'])
warnings
```

## Frames at each warning

```python
import cv2
import matplotlib.pyplot as plt

cap = cv2.VideoCapture(str(outputs['result']))
frames = list(warnings['frame'][:4])
fig, axes = plt.subplots(1, max(len(frames), 1), figsize=(6 * max(len(frames), 1), 4), squeeze=False)
for ax, frame in zip(axes[0], frames):
    cap.set(cv2.CAP_PROP_POS_FRAMES, frame)
    ok, img = cap.read()
    if ok:
        ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    ax.set_title(f'frame {frame}')
    ax.axis('off')
cap.release()
plt.tight_layout()
plt.show()
```
